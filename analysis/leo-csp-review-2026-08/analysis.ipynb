## Context & Methods
### Key Assumptions
Screenshot fills are authoritative. Same-contract Moomoo snapshots within 20 minutes are observed evidence; otherwise IV is solved from the fill with a Black–Scholes model using local underlying snapshots, 4% risk-free rate, and bounded dividend-yield assumptions. TSM remains unavailable because its underlying was not archived.


```python
import json
from pathlib import Path
results = json.loads(Path('results.json').read_text(encoding='utf-8'))
len(results['entries']), len(results['exits'])
```

## Data
The bounded entry and exit records used in the report.


```python
[(row['contract'], row['greeks_status'], row['delta'], row['iv']) for row in results['entries']]
```

## Results
Recompute the headline portfolio and exit statistics from the serialized rows.


```python
entries = results['entries']
exits = [row for row in results['exits'] if row['entry_visible']]
visible_collateral = sum(row['collateral'] for row in entries)
visible_premium = sum(row['gross_premium'] for row in entries)
weighted_entry_yield = visible_premium / visible_collateral
weighted_capture = sum((row['entry_credit'] - row['exit_debit']) * 100 for row in exits) / sum(row['entry_credit'] * 100 for row in exits)
visible_collateral, weighted_entry_yield, weighted_capture
```

## Takeaways
The visible winner sample supports an aggressive, higher-Delta CSP style and near-full premium capture. It does not establish a true annual account return because losing or still-open trades are not shown.
